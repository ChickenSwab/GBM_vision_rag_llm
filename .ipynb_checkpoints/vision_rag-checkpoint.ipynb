```python
import json
import pickle
import faiss
import numpy as np

from sentence_transformers import SentenceTransformer
```

```python
with open(
        "vision_output.json",
        "r"
) as f:

    vision_data = json.load(f)

features = vision_data["features"]
query = vision_data["query"]
age = vision_data["age"]
gender = vision_data["gender"]

print(query)
```

```python
MODEL_NAME = "BAAI/bge-large-en-v1.5"

encoder = SentenceTransformer(
    MODEL_NAME
)
```

```python
index = faiss.read_index(
    "cgga.index"
)

with open(
        "documents.pkl",
        "rb"
) as f:

    documents = pickle.load(f)

with open(
        "metadata.pkl",
        "rb"
) as f:

    metadata = pickle.load(f)
```

```python
query_embedding = encoder.encode(
    query,
    normalize_embeddings=True
)

query_embedding = np.asarray(
    query_embedding,
    dtype=np.float32
)
```

```python
scores, indices = index.search(
    query_embedding.reshape(1, -1),
    20
)
```

```python
query_metadata = {

    "histology": "GBM",

    "grade": "WHO IV",

    "idh": "Wildtype",

    "mgmt": "methylated",

    "1p19q": "Non-codel",

    "age": age,

    "gender": gender
}
```

```python
reranked = rerank_candidates(
    scores,
    indices,
    metadata,
    documents,
    query_metadata
)
```

```python
for rank, result in enumerate(
        reranked[:5],
        start=1):

    print("="*80)

    print(
        f"Rank {rank}"
    )

    print(
        "Semantic score:",
        result["semantic_score"]
    )

    print(
        "Final score:",
        result["final_score"]
    )

    print()

    print(
        result["metadata"]
    )

    print()

    print(
        result["document"]
    )
```
